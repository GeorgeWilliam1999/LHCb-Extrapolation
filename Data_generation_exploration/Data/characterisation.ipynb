# Training-sample characterisation

Loads the results produced by `harvest_states.py` -> `make_training_set.py` -> `characterise.py` (see README.md for the full argument). This notebook is the analysis surface: rerun the scripts, then rerun this.

```python
import json, numpy as np, pandas as pd
from IPython.display import Image, display

tr = np.load('training_v1/train_mb100_v1.npz')
meta = json.load(open('training_v1/train_mb100_v1.meta.json'))
print('rows:', len(tr['X']), ' per leg:', meta['rows']['per_leg'])
print('split:', meta['rows']['split'])
print('core rows:', int(tr['CORE'].sum()), '(%.1f%%)' % (100*tr['CORE'].mean()))
```

```python
# the integrity gates
print(json.dumps(meta['gates'], indent=1))
```
